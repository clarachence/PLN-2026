```python
# Instalação e download de recursos (rodar 1x)
!pip install -q nltk spacy pandas
!python -m spacy download pt_core_news_md -q

import nltk
import spacy
import pandas as pd

nltk.download('stopwords', quiet=True)
nltk.download('rslp', quiet=True)

nlp = spacy.load("pt_core_news_md")

print("Ambiente configurado com sucesso.")
```

```python
frase_exercicio = "O proprietário alugou a casa para a família rapidamente."
doc_exercicio = nlp(frase_exercicio)
```

# Passo 1: Tokenização

```python
tokens = [token.text for token in doc_exercicio]
print("Tokens:", tokens)
```

# Passo 2: Remoção de pontuação e stopwords

```python
tokens_limpos = [token.text for token in doc_exercicio if not token.is_punct and not token.is_stop]
print("Tokens limpos:", tokens_limpos)
```

# Passo 3: Stemming e Lematização (tabela comparativa)

```python
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
stems = [stemmer.stem(t) for t in tokens_limpos]

doc_limpo = nlp(" ".join(tokens_limpos))
lemas = [token.lemma_ for token in doc_limpo]

comparacao = pd.DataFrame({
    "Token": tokens_limpos,
    "Stemming (RSLP)": stems,
    "Lematização (spaCy)": lemas
})
comparacao
```

# Passo 4: POS Tagging

```python
dados_pos = [{"Token": t.text, "POS": t.pos_} for t in doc_exercicio]
pd.DataFrame(dados_pos)
```

# Passo 5: Sintagma Nominal e Sintagma Verbal

```python
sintagmas_nominais = [chunk.text for chunk in doc_exercicio.noun_chunks]

verbo = next(token for token in doc_exercicio if token.pos_ == "VERB")

tokens_vp = [
    token for token in doc_exercicio
    if token.i >= verbo.i and not token.is_punct
]

sv_texto = " ".join(token.text for token in tokens_vp)

print("Sintagmas Nominais (NP):", sintagmas_nominais)
print("Sintagma Verbal (VP):", sv_texto)
```

# Passo 6: Similaridade entre duas palavras da frase

```python
palavra_a = nlp("proprietário")
palavra_b = nlp("casa")

print(f"Similaridade proprietário x casa: {palavra_a.similarity(palavra_b):.3f}")
```
