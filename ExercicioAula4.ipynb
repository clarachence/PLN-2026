# Preparação do ambiente

```python
!pip install -q spacy pandas
!python -m spacy download pt_core_news_sm -q

import spacy
import pandas as pd
from spacy import displacy

nlp = spacy.load("pt_core_news_sm")

frase_exercicio = "O corretor apresentou o imóvel para o cliente ontem."
doc = nlp(frase_exercicio)

print("Frase analisada:", frase_exercicio)
```

# 1. POS Tagging

```python
classes = {
    "DET": "Determinante / artigo",
    "NOUN": "Substantivo",
    "VERB": "Verbo",
    "ADP": "Preposição",
    "ADV": "Advérbio",
    "PUNCT": "Pontuação"
}

tabela_pos = pd.DataFrame({
    "Token": [token.text for token in doc],
    "POS": [token.pos_ for token in doc],
    "Classe gramatical": [classes.get(token.pos_, token.pos_) for token in doc]
})

display(tabela_pos)
```

# 2. Visualização com displacy

```python
displacy.render(
    doc,
    style="dep",
    jupyter=True,
    options={"distance": 100, "collapse_punct": False}
)
```

# 3. Extração do SN e do SV

```python
# Encontrar a raiz
raiz = [token for token in doc if token.dep_ == "ROOT"][0]
print("Raiz da frase:", raiz.text)

# Encontrar o sujeito nominal
sujeito = [token for token in doc if token.dep_ == "nsubj"][0]
print("Sujeito:", sujeito.text)

# O subtree do sujeito inclui o núcleo e os elementos ligados a ele
sintagma_nominal = list(sujeito.subtree)
print("Sintagma Nominal (SN):", [token.text for token in sintagma_nominal])

# O subtree da raiz contém a estrutura verbal completa
# Retiramos dele os tokens que pertencem ao SN
sintagma_verbal = [
    token for token in raiz.subtree
    if token not in sintagma_nominal
]
print("Sintagma Verbal (SV):", [token.text for token in sintagma_verbal])
```

```python
sn_texto = " ".join(token.text for token in sintagma_nominal)
sv_texto = " ".join(token.text for token in sintagma_verbal)

print("SN =", sn_texto)
print("SV =", sv_texto)
```
