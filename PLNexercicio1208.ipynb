```python
!pip install -q spacy nltk pandas
!python -m spacy download pt_core_news_sm -q

import nltk
import spacy
import pandas as pd

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

```python
frase_exercicio = "Olá! Gostaria de agendar uma visita para ver o imóvel de 3 quartos amanhã às 14h?"
```

```python
# Tokenização com spaCy
doc_spacy = nlp(frase_exercicio)
tokens_spacy = [token.text for token in doc_spacy]
print("Tokens com spaCy:", tokens_spacy)
```

```python
# Remoção de pontuação
tokens_sem_pontuacao = [token.text for token in doc_spacy if not token.is_punct]
print("Tokens sem pontuação:", tokens_sem_pontuacao)
```

```python
# Contagem: original vs. sem pontuação

print("Quantidade de tokens originais:", len(tokens_spacy))
print("Quantidade de tokens sem pontuação:", len(tokens_sem_pontuacao))
```
